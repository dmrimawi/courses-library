## 4. First challenge: estimate a delivery time

A customer places a food order. Before dispatch, we want an estimated arrival time.

- **Handwritten rule:** 8 minutes + 4 minutes per kilometer.
- **Problem:** Traffic can make equally long trips take different amounts of time.
- **ML idea:** Use examples of past deliveries to **learn** a relationship.

We will compare the two approaches, then see whether our model deserves trust.

### What information do we have?

| Delivery | Distance (km) | Traffic | Actual time (min) |
|:---:|---:|---|---:|
| D01 | 1.1 | Light | 10 |
| D02 | 1.8 | Moderate | 20 |
| D03 | 2.0 | Light | 13 |
| D04 | 2.4 | Heavy | 27 |

**Ask before predicting:** Which columns are known **before** dispatch? Which value is only known **after** the delivery?

### Learn the vocabulary from our table

- **Dataset:** the complete collection of delivery records.
- **Observation:** one delivery, or one row.
- **Feature:** information used to predict, such as `distance_km`.
- **Target:** what we want to predict—here, `actual_minutes`.

Predicting minutes is a **regression** task: the output is a number.

**Important:** Never give the model the actual completion time as an input when predicting that same time! It is not known before dispatch. Using future information is **data leakage**.

### Let's see our data in Python

- `pandas` manages tables, and `DataFrame(...)` builds one.
- `.head(8)` displays the first eight observations.
- We will learn this syntax properly in **L03**. Today, **follow the data, not the details of each command**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# We use explicit inputs and completed-delivery outcomes.
delivery_ids = [f"D{i:02d}" for i in range(1, 29)]
distances = [1.1,1.8,2.0,2.4,3.1,3.4,4.0,4.3,4.8,5.2,5.7,6.0,6.4,6.9,
             7.3,7.6,8.1,8.5,9.0,9.3,2.7,4.5,5.5,7.0,8.8,3.8,6.6,9.7]
traffic_scores = [0,1,0,2,1,0,2,1,0,1,2,0,1,2,
                  0,1,2,0,1,2,2,0,1,0,2,1,1,0]
# Values are teaching examples, not measurements from real deliveries.
actual_times = [10.0,20.0,13.0,27.0,25.0,18.0,36.0,32.0,24.0,34.0,41.0,30.0,
                37.0,46.0,33.0,40.0,50.0,37.0,46.0,53.0,31.0,24.0,33.0,33.0,
                54.0,28.0,40.0,41.0]

deliveries = pd.DataFrame({
    "delivery_id": delivery_ids,
    "distance_km": distances,
    "traffic_score": traffic_scores,   # 0 = light, 1 = moderate, 2 = heavy
    "actual_minutes": actual_times,
})
deliveries.head(8)
```

**Read the output:** `delivery_id` identifies each delivery; distance and traffic describe conditions **before dispatch**; actual minutes are recorded **afterward**.

Next, `.shape` tells us the number of rows and columns; `.min()` and `.max()` show the smallest and largest distances.

```python
print("Rows and columns:", deliveries.shape)
print("Shortest distance (km):", deliveries["distance_km"].min())
print("Shortest distance (km):", np.min(deliveries["distance_km"]))
print("Longest distance (km):", deliveries["distance_km"].max())
print("Longest distance (km):", np.max(deliveries["distance_km"]))
print("Missing entries in each column:")
print(deliveries.isna().sum())
```

### Before learning, look for a pattern

- Our complete table contains **28 deliveries** and **4 columns**.
- The chart plots **distance** (x-axis) against **actual minutes** (y-axis).
- Colors show **light / moderate / heavy traffic**.


```python
traffic_names = {0: "Light traffic", 1: "Moderate traffic", 2: "Heavy traffic"}
traffic_colors = {0: "#2468a0", 1: "#e69f00", 2: "#b04452"}

fig, ax = plt.subplots(figsize=(9, 5))
for score in [0, 1, 2]:
    group = deliveries[deliveries["traffic_score"] == score]
    ax.scatter(group["distance_km"], group["actual_minutes"],
               label=traffic_names[score], color=traffic_colors[score],
               s=75, edgecolors="white", linewidth=0.7, alpha=0.9)
ax.set(title="Previous deliveries: distance, traffic, and actual time",
       xlabel="Distance available at dispatch (km)",
       ylabel="Actual completed-delivery time (minutes)")
ax.grid(alpha=0.18)
ax.legend(title="Traffic observed at dispatch")
plt.show()
```

**Notice:** Longer trips often take longer, but similar distances have different outcomes. The dots suggest a pattern—not a perfect rule or proof of cause.

### First attempt: a fixed rule

We'll program **8 + 4 × distance**. It never adjusts its numbers using earlier deliveries.



*The code that follows starts with the handwritten rule; later we fit the learned rule.*

```python
def fixed_rule(distance_km):
    return 8 + 4 * distance_km

for km in [2, 5, 8]:
    print(f"Distance {km} km -> rule predicts {fixed_rule(km):.1f} minutes")
```

**The rule works mechanically:** for 2, 5, and 8 km it gives 16, 28, and 40 minutes. But are those numbers good estimates?

### A fairer check: separate learning and testing

- **Training set:** 21 deliveries used to learn a relationship.
- **Test set:** 7 different deliveries used to check predictions.

**Do not train on the test outcomes.**

Today's split is fixed for reproducibility. Real projects need carefully chosen test data that resemble future situations.

```python
# These IDs are kept out of the model's learning step.
test_ids = {"D04", "D08", "D12", "D16", "D20", "D24", "D27"}
training = deliveries[~deliveries["delivery_id"].isin(test_ids)].copy()
testing = deliveries[deliveries["delivery_id"].isin(test_ids)].copy()

print("Training observations:", len(training))
print("Held-out test observations:", len(testing))
print("Test delivery IDs:", list(testing["delivery_id"]))
```

### Train the model

**Linear regression** learns a line relating distance to delivery time. We study its mathematics in **L11**.

- `LinearRegression()` creates a model.
- `X_train` contains **input distances**; `Y_train` contains **known times**.
- `.fit(X_train, Y_train)` lets the model **learn from training examples**.

**For now, we use distance alone.** Later we'll see what happens when we include traffic.

```python
X_train = training[["distance_km"]]   # Input feature available before delivery
Y_train = training["actual_minutes"]  # Outcome measured afterward

model = LinearRegression()             # Create an untrained model
model.fit(X_train, Y_train)            # Learn from training examples only

print(f"Learned starting point: {model.intercept_:.2f} minutes")
print(f"Learned minutes per km: {model.coef_[0]:.2f}")
# 8 + 4 x Distance
# 11.35 + 3.94 x Dstance 
# w1 x X1 + w2 x X2 + .... + b + w5 x minutes = minutes
```

### Our first learned prediction

The model found a relationship that looks like:

$$\widehat{\mathrm{minutes}} = a + b\times\mathrm{distance}.$$

- **a and b** are numbers estimated from the training examples.
- The hat means **predicted** time, not actual time.

**New order:** 5 km. Can our model estimate the duration?

`.predict(...)` returns an estimated time for an input with the **same feature** used during training.

**Before running:** Do you expect the same result as our fixed **28-minute** rule?

```python
new_order = pd.DataFrame({"distance_km": [5.0]})
new_prediction = model.predict(new_order)[0]
print(f"Handwritten rule: {fixed_rule(5.0):.1f} minutes")
print(f"Learned model:     {new_prediction:.1f} minutes")
```

---

### Can we *trust* it? 

We need to check the errors.

**A prediction is not a fact.** It is an estimate for a trip that hasn't finished.

### How good is the model on deliveries it never trained on?

We'll compare its predictions against the **seven held-out actual times**.

- **Error:** prediction minus actual duration.
- **Absolute error:** how many minutes we missed, ignoring the sign.

```python
X_test = testing[["distance_km"]]
testing["predicted_minutes"] = model.predict(X_test)
testing["rule_minutes"] = testing["distance_km"].apply(fixed_rule)
testing["signed_error"] = testing["predicted_minutes"] - testing["actual_minutes"]
testing["absolute_error"] = testing["signed_error"].abs()
testing["signed_rules_error"] = testing["rule_minutes"] - testing["actual_minutes"]
testing["absolute_rules_error"] = testing["signed_rules_error"].abs()

testing[["delivery_id", "distance_km", "traffic_score", "actual_minutes", "rule_minutes",
         "predicted_minutes", "absolute_rules_error", "absolute_error"]].round(1)
```

### Measure the size of our mistakes

A prediction of 25 minutes for a trip that took 31 minutes has an **absolute error of 6 minutes**.

**Mean absolute error (MAE)** = average absolute miss, measured in minutes. **Smaller means a closer fit on this particular test set.**

Now compare our **learned model** with the **fixed rule**, using the same held-out deliveries.

```python
model_mae = testing["absolute_error"].mean()
rule_mae = (testing["absolute_rules_error"]).mean()
print(f"Distance-only learned model: {model_mae:.1f} minutes MAE on held-out deliveries")
print(f"Handwritten rule:            {rule_mae:.1f} minutes MAE on held-out deliveries")
```

**Interpret carefully:** Lower test MAE means smaller errors **on these examples**, not a guarantee about every future delivery.

### Visualize the learned rule

- **Circles:** training deliveries.
- **Diamonds:** unseen test deliveries.
- **Black line:** the model's prediction from distance alone.

```python
line_distances = np.linspace(1, 10, 100)  # 100 distances for drawing a smooth line
line_table = pd.DataFrame({"distance_km": line_distances})
line_predictions = model.predict(line_table)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(training["distance_km"], training["actual_minutes"],
           label="Training deliveries", color="#3777a7", s=65, alpha=0.8)
ax.scatter(testing["distance_km"], testing["actual_minutes"],
           label="Held-out deliveries", color="#d18d36", marker="D", s=90)
ax.plot(line_distances, line_predictions, color="#242424", linewidth=2.5,
        label="Learned prediction from distance")
ax.set(title="A learned line: useful trend, imperfect predictions",
       xlabel="Distance at dispatch (km)",
       ylabel="Delivery time (minutes)")
ax.grid(alpha=0.15)
ax.legend()
plt.show()
```

**Notice:** Some points fall above the line; others fall below. A model can capture a useful **trend** without predicting every example perfectly.

**The question for later:** Would traffic improve these predictions?

### Data Science, Machine Learning, and Data Mining

- **Data Science:** asking useful questions and collecting, cleaning, exploring, analyzing, and communicating data.
- **Machine Learning:** building models that **learn from examples or experience** to make predictions or improve behavior.
- **Data Mining:** discovering useful patterns and relationships in datasets.

They overlap, but they are **not synonyms**.



### Fixed rules or learned patterns?

- **Traditional rules:** *We* choose the instructions (for example, `8 + 4 × km`).
- **Learning from examples:** A model adjusts its parameters using previous outcomes.

Neither wins automatically. We check which works for the real task.

### Learning is not the same as memorizing

A useful model should recognize relationships that also help with **new examples**.

That is why we kept some deliveries aside for **testing**.

### Why predictions can fail

- **Missing information:** traffic, weather, or road closures might matter.
- **Bad data:** wrong units or inaccurate recorded times.
- **Changing conditions:** future deliveries may differ from old ones.
- **Model limits:** a straight line cannot describe everything.

**One takeaway:** Always test on appropriate new data before trusting predictions.


### AI, ML, and Deep Learning: related, not identical

- **Artificial Intelligence (AI):** the broad goal of building systems that carry out tasks associated with intelligence.
- **Machine Learning (ML):** a major approach within AI—systems learn from data or interaction.
- **Deep Learning (DL):** ML using multilayer neural networks.

**Data Science overlaps with ML** but also includes tasks that do not involve training a model.

**Remember:** Deep learning is a family of models; it is **not** the only way to do ML.

## 6. Different goals, different learning approaches

### Supervised learning — outcomes are known

We learn from examples with a **target** (such as the recorded delivery time).

- **Regression (L11–L12):** predict a **number**, e.g., delivery duration.
- **Classification (L13–L18):** predict a **category**, e.g., *late* vs. *on time*.

**Key idea:** The training examples include the answers.

### Unsupervised learning — outcomes are not labeled

We look for structure **without supplying the correct group or outcome**.

- **Clustering (L19–L20):** group deliveries with similar patterns.
- **Anomaly detection (L21):** flag unusually different trips for investigation.

**Key idea:** Finding a group and spotting an unusual case are different questions.

### Limited labels — two different strategies (L22)

- **Semi-supervised:** learn using a few **labeled** records together with many **unlabeled** ones.
- **Active learning:** select **which records a person should label next** to use limited labeling time wisely.

Both help when labels are costly, but they solve different problems.

### Reinforcement learning (L23–L24)

An **agent** takes actions in an **environment** and receives **rewards**. It learns which actions tend to work well.

**Example:** A small game agent learns how to move toward a goal.

### Neural networks and deep learning (L25–L30)

- A **neural network** is a model family made of connected computational units.
- Some neural networks learn from labeled examples; others can be trained in other settings.
- **Deep learning** uses networks with multiple layers.

**Not a separate category of training feedback:** Neural networks are *models*, not another kind of supervision.



### Quick guide: Which question are we asking?

- Known correct outcomes? → **Supervised learning**.
- No known outcomes; looking for groups or unusual cases? → **Unsupervised learning**.
- Only a few labels? → **Semi-supervised**.
- Actions with rewards? → **Reinforcement learning**.

We will study all of these more carefully in their dedicated lectures.

## 7. One delivery dataset, four different questions

**Changing the question changes what we need to learn.**

### A quick classification preview

We can label historical trips **late** or **on time** by comparing their actual duration with a promised duration.

**Important:** This is just a way to *construct historical labels*, **not** a trained classifier. We will learn classification in **L13**.



*The four panels pose different questions. The same feature table does not make them equivalent tasks.*

```python
deliveries["promised_minutes"] = (
    9 + 3.5 * deliveries["distance_km"] + 6 * deliveries["traffic_score"]
)
deliveries["late"] = deliveries["actual_minutes"] > deliveries["promised_minutes"]

print("Counts of completed deliveries by historical late/on-time status:")
print(deliveries["late"].map({False: "On time", True: "Late"}).value_counts())
deliveries[["delivery_id", "distance_km", "traffic_score",
            "promised_minutes", "actual_minutes", "late"]].head(10).round(1)
```

### Read two views of the same deliveries

- **Left chart:** Color shows **traffic levels**.
- **Right chart:** Color shows **late vs. on time**.

Both plot distance against historical actual time. **Same points; different question.**

**Caution:** We use actual durations to **inspect historical records**. If predicting *before* a delivery, actual duration cannot be an input.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.7), sharex=True, sharey=True)
for state, name, color in [(False, "On time", "#2b7a78"), (True, "Late", "#b44a52")]:
    subset = deliveries[deliveries["late"] == state]
    axes[0].scatter(subset["distance_km"], subset["actual_minutes"],
                    color=color, s=70, alpha=.86, label=name)
for traffic in [0,1,2]:
    subset = deliveries[deliveries["traffic_score"] == traffic]
    axes[1].scatter(subset["distance_km"], subset["actual_minutes"],
                    color=traffic_colors[traffic], s=70, alpha=.86,
                    label=traffic_names[traffic])
for ax, title in zip(axes, ["Known outcome: was it late?", "Describing trips by known traffic"]):
    ax.set(title=title, xlabel="Distance (km)", ylabel="Actual minutes")
    ax.grid(alpha=.12)
    ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
```

### What would count as an anomaly?

A **very slow delivery** might be unusual—but perhaps heavy traffic explains it.

- Unusual **compared with what**?
- Would you investigate a 40-minute trip of 2 km or 10 km first?

**Anomaly detection (L21)** looks for cases deserving attention, not automatic proof of mistakes.

**Key idea:** Start by defining the question; then choose a learning approach that matches it.

### Activity 1 · Add traffic 

Our first model uses **distance only**. But our plots suggest traffic matters.

**Before running:** If we add `traffic_score`, should test errors improve? Could they ever get worse?

We'll use the **same training and test deliveries** to compare fairly.

The new model will use **distance + traffic** as inputs. The next cell reports its **test MAE**, alongside the original model's test MAE.

**Interpret:** A lower error here supports the idea that traffic helps *on this small test set*. It does not guarantee the same result for all future trips.

```python
more_features = ["distance_km", "traffic_score"]
model_with_traffic = LinearRegression()
model_with_traffic.fit(training[more_features], training["actual_minutes"])

two_feature_predictions = model_with_traffic.predict(testing[more_features])
two_feature_mae = np.abs(two_feature_predictions - testing["actual_minutes"]).mean()

print(f"Distance-only test MAE:      {model_mae:.2f} minutes")
print(f"Distance + traffic test MAE: {two_feature_mae:.2f} minutes")
print(f"Difference (old - new):      {model_mae - two_feature_mae:.2f} minutes")
```
