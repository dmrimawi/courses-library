### 1.1 Why is Machine Learning worth learning?

Think about familiar services: a map estimates travel time, a shop recommends useful items, an email app moves suspicious messages to spam, and a city may monitor sensors for unusual readings. These services deal with repeated decisions in situations that change. The computer needs to act on information, but the exact answer is not always obvious in advance.

A person could write rules for every situation. For instance, *add five minutes whenever there is heavy traffic*. Sometimes such rules work well. But many problems have interacting factors, and we may not know the right size of each effect. A different possibility is to look at what happened before and learn a relationship from those examples.

A learned relationship is not a guarantee. The world can change, the examples can be incomplete, and recorded data can contain mistakes. Our aim is not only to make a number appear on the screen, but to understand **when that number is useful and when it might mislead us**.

Today’s Machine Learning systems did not appear suddenly. The story began with a question about **intelligent behavior**. Researchers then explored how a computer might improve through **experience**, even when nobody could write every decision rule.

The timeline below follows a few turning points rather than every discovery in the field.



*The field advanced through several interacting ideas—not one method replacing all previous methods.*

### 1950 — Alan Turing: can a machine appear intelligent?

- **Development:** In *Computing Machinery and Intelligence*, Alan Turing discusses the **imitation game**, a way to reason about observable intelligent behavior.
- **Why it matters:** It asks what a machine might *do*, instead of assuming we already have a simple definition of intelligence.
- **Important distinction:** Turing was discussing machine intelligence, **not** introducing modern supervised learning.

**Connection:** This question inspired research into machine intelligence. Learning from experience became one possible way to pursue it.

### 1957–1959 — The idea becomes a learning program

**1957–1958 | Frank Rosenblatt and the perceptron**

- **Development:** A simple computational model adjusts its behavior using examples; early versions were demonstrated as physical systems.
- **Why it matters:** It is an early building block of neural-network learning. We return to the **perceptron in L25**.

**1959 | Arthur Samuel and checkers**

- **Development:** Samuel publishes research on a checkers program that improved through experience.
- **Why it matters:** Instead of writing every good move in advance, researchers could design programs that improved using evidence from play.

**Connection:** A machine can adjust its behavior, but how do we find reliable patterns and check whether they work on new cases?

### 1980s–1990s — Learning methods become more practical

- **1986 | Backpropagation research:** Influential work shows how multi-layer networks can adjust their internal parameters. We will build toward this in **L26–L28**.
- **1995 | Support-vector networks:** One example of increasingly powerful statistical methods for finding decision boundaries. We will meet **SVMs in L18**.
- **Why these developments matter:** Learning algorithms became better at turning examples into relationships that can be checked on unfamiliar data.

**Connection:** Useful learning methods need appropriate data and enough computing power to train them.

### 2012 → 2017 → 2020s — Learning at a much larger scale

- **2012 | Deep learning and ImageNet:** A deep neural network achieves a breakthrough result in a major image-recognition competition. **Large labeled datasets** and strong computing hardware play an important role.
- **2017 | Transformers:** *Attention Is All You Need* introduces an architecture that shapes later language and sequence-processing research.
- **2020s | Generative AI:** Systems that produce **text, images, and code** become widely visible. Generated content still needs careful checking.

**Why it matters:** Progress came from a combination of **data, learning methods, and computing resources**—not one invention alone.

> **History takeaway:** Newer is not always better for every task. Regression, decision trees, and other simpler approaches remain useful. We choose a method because it fits the **problem**, not because it has the newest name.

## 4. Our first real problem: can we estimate a delivery time?

Imagine a food-delivery service. A customer places an order. Before a courier leaves, the service wants to estimate how many minutes the trip will take.

A simple rule might be:

> Start with 8 minutes and add 4 minutes for each kilometer.

This is attractive because everyone can understand it. A 3 km trip would take **20 minutes** according to the rule. But two deliveries of the same distance can take very different amounts of time. Heavy traffic, road conditions, and weather may matter, and our numbers might be wrong for the city or time of day.

Instead, suppose we have records of **previous completed deliveries**. We can observe the distance and the level of traffic known at dispatch time, together with the actual time recorded *after* each delivery. That historical record may help us discover a useful relationship.

Today's table is **synthetic** (constructed for teaching, not collected from real people). The values are deliberately small and realistic enough to reason about. They do not represent the performance of any actual delivery company.

### 4.1 The first six records

| Delivery | Distance at dispatch (km) | Traffic at dispatch | Actual time after delivery (min) |
|---|---:|---|---:|
| D01 | 1.1 | light | 10.0 |
| D02 | 1.8 | moderate | 20.0 |
| D03 | 2.0 | light | 13.0 |
| D04 | 2.4 | heavy | 27.0 |
| D05 | 3.1 | moderate | 25.0 |
| D06 | 3.4 | light | 18.0 |

Notice that D03 and D04 are similar in distance but very different in traffic and duration. It is a small warning against relying on distance alone.

**Before running code:** Which columns would be known *before* the courier starts? Which column cannot be known until *after* the trip? That distinction will determine which information is safe to use for a real prediction.

### 4.2 Four names for things we already understand

A **dataset** is a collection of related observations. Here it is a small table of delivery records.

An **observation** is one recorded case. Each row refers to one delivery. Some fields identify it; others describe it.

A **feature** is information the model may use as an input to make a prediction. We will first use `distance_km`. Later, we will also consider `traffic_score` as an input. A delivery ID is not a meaningful predictive feature here: it simply identifies a record.

A **target** (or **label** in the broad supervised-learning sense) is the outcome we want to predict. Our target is `actual_minutes`. Since it is a numerical duration, this will become a **regression** problem. In classification, the target is a category such as “late” or “on time.”

We have a strict timing rule: an input must be available **at the moment the prediction is made**. `actual_minutes` is a target, never an input for predicting that same delivery time. Including it among the inputs would give the model information from the future, a form of **data leakage** that makes a demonstration look more accurate than a real system could be.

### 4.3 Enter Python: an embedded dataset

We will now construct the full table. Only four Python tools matter at this stage:

- `pandas` helps us represent and inspect tabular data. We will learn it carefully in **L03**.
- `DataFrame(...)` turns named columns into a table. Think of it as a spreadsheet stored in Python.
- A **list** (values between square brackets) holds several entries in a sequence.
- `.head(8)` shows the first eight rows rather than printing the whole table.

There is **no need to master these commands today**. For now, follow the story told by the table. All values are written inside the notebook; it works offline.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# We use explicit inputs and completed-delivery outcomes.
delivery_ids = [f"D{i:02d}" for i in range(1, 29)]
distances = [1.1,1.8,2.0,2.4,3.1,3.4,4.0,4.3,4.8,5.2,5.7,6.0,6.4,6.9,
             7.3,7.6,8.1,8.5,9.0,9.3,2.7,4.5,5.5,7.0,8.8,3.8,6.6,9.7]
traffic_scores = [0,1,0,2,1,0,2,1,0,1,2,0,1,2,
                  0,1,2,0,1,2,2,0,1,0,2,1,1,0]
# Values are teaching examples, not measurements from real deliveries.
actual_times = [10.0,20.0,13.0,27.0,25.0,18.0,36.0,32.0,24.0,34.0,41.0,30.0,
                37.0,46.0,33.0,40.0,50.0,37.0,46.0,53.0,31.0,24.0,33.0,33.0,
                54.0,28.0,40.0,41.0]

deliveries = pd.DataFrame({
    "delivery_id": delivery_ids,
    "distance_km": distances,
    "traffic_score": traffic_scores,   # 0 = light, 1 = moderate, 2 = heavy
    "actual_minutes": actual_times,
})
deliveries.head(8)
````

The result is a table with **four columns**. `delivery_id` is an identifier; `distance_km` and `traffic_score` were available before each trip; `actual_minutes` was measured when the delivery finished. The first rows match our example table.

Traffic is written as 0, 1, and 2 only so it is easy to use in a demonstration. The numbers are **codes for ordered categories**, not proof that “heavy” traffic is exactly twice “moderate” traffic. Working with categories carefully belongs to our later data lessons.

Let's check two simple facts about our complete dataset. `.shape` tells us how many rows and columns the table contains, and `.min()` and `.max()` find the smallest and largest recorded distances.

````python
print("Rows and columns:", deliveries.shape)
print("Shortest distance (km):", deliveries["distance_km"].min())
print("Longest distance (km):", deliveries["distance_km"].max())
print("Missing entries in each column:")
print(deliveries.isna().sum())
````

We have **28 observations and 4 columns**, with distances from **1.1 km to 9.7 km**, and no missing entries in this constructed table. That is a quick inspection, **not** proof that data are correct or representative. Real data may contain duplicate deliveries, wrong times, unexpected units, or missing weather information. We will learn to investigate those problems in **L07**.

### 4.4 Turn the observations into a picture

Before asking a model to learn, it helps to look. The next chart places **distance on the horizontal axis** and **actual delivery duration on the vertical axis**. Every point represents one completed delivery. The point color represents a traffic category: blue for light, orange for moderate, and red for heavy.

**Predict before running:** Will longer trips always take longer? Do you expect all dots to lie on one perfect line?

````python
traffic_names = {0: "Light traffic", 1: "Moderate traffic", 2: "Heavy traffic"}
traffic_colors = {0: "#2468a0", 1: "#e69f00", 2: "#b04452"}

fig, ax = plt.subplots(figsize=(9, 5))
for score in [0, 1, 2]:
    group = deliveries[deliveries["traffic_score"] == score]
    ax.scatter(group["distance_km"], group["actual_minutes"],
               label=traffic_names[score], color=traffic_colors[score],
               s=75, edgecolors="white", linewidth=0.7, alpha=0.9)
ax.set(title="Previous deliveries: distance, traffic, and actual time",
       xlabel="Distance available at dispatch (km)",
       ylabel="Actual completed-delivery time (minutes)")
ax.grid(alpha=0.18)
ax.legend(title="Traffic observed at dispatch")
plt.show()
````

**Read the chart:** The overall trend rises: longer distances are often associated with longer times. But points at similar distances are not identical. Colored points suggest that traffic may explain some differences. The graph helps us notice **a pattern**, but by itself it does not show that distance *causes* every change in delivery duration.

Also, there are only 28 invented examples, within a narrow distance range. The chart does not tell us what will happen on a 50 km trip or after a major road closure.

### 4.5 First attempt: a rule that does not learn

Our original handwritten rule was **8 minutes + 4 minutes per kilometer**. The function below implements exactly that decision. `def` gives the function a name; the value inside parentheses is the distance we supply. No past example is used to adjust the numbers 8 or 4.



*The code that follows starts with the handwritten rule; later we fit the learned rule.*

````python
def fixed_rule(distance_km):
    return 8 + 4 * distance_km

for km in [2, 5, 8]:
    print(f"Distance {km} km -> rule predicts {fixed_rule(km):.1f} minutes")
````

The rule predicts **16, 28, and 40 minutes** for 2, 5, and 8 km respectively. It always gives the same answer for the same distance, even if traffic changes. This does not make rule-based approaches “bad”: they are often fast, understandable, and appropriate. The question is whether a rule written by hand captures enough of the real variation for our purpose.

### 4.6 Separate examples for learning and checking

To find out whether a model can help with **future deliveries**, we should check it on examples that were *not* used during learning. We will place 21 records in a **training set** and keep 7 separate records in a **test set**.

A **training set** is the part the model sees while learning. A **test set** is held aside until we want to check how it handles unfamiliar examples. A test record can have a known actual time in our historical table, but the model must not see that actual time during training.

This is a fixed illustrative split, chosen so today's result stays reproducible. In a real deployment we would think carefully about time order, changing road conditions, and how the test period represents future requests. Later lectures will make model evaluation much more rigorous.

````python
# These IDs are kept out of the model's learning step.
test_ids = {"D04", "D08", "D12", "D16", "D20", "D24", "D27"}
training = deliveries[~deliveries["delivery_id"].isin(test_ids)].copy()
testing = deliveries[deliveries["delivery_id"].isin(test_ids)].copy()

print("Training observations:", len(training))
print("Held-out test observations:", len(testing))
print("Test delivery IDs:", list(testing["delivery_id"]))
````

The printed counts should be **21 for training** and **7 for testing**. The symbols in the first two lines are pandas filtering operations. You do not need to remember their syntax now; what matters is that the two groups do **not** overlap.

### 4.7 Let a simple model learn a relationship

We will use **linear regression**, a method we will study carefully in **L11**. The idea today is simple: from examples, it learns a line that approximately connects the input (distance) with the output (minutes). Unlike our fixed rule, the line's numbers are estimated from training examples.

We need just four commands:

- `LinearRegression()` creates a new, initially untrained model.
- `training[["distance_km"]]` selects the input feature as a two-dimensional table (as this library expects).
- `training["actual_minutes"]` selects the known training targets.
- `.fit(inputs, targets)` is the **training** step: it finds values for the model's line.

The model learns from **distance alone** on purpose. We will soon see an important consequence of leaving out traffic.

````python
X_train = training[["distance_km"]]   # Input feature available before delivery
Y_train = training["actual_minutes"]  # Outcome measured afterward

model = LinearRegression()             # Create an untrained model
model.fit(X_train, Y_train)            # Learn from training examples only

print(f"Learned starting point: {model.intercept_:.2f} minutes")
print(f"Learned minutes per km: {model.coef_[0]:.2f}")
````

These two learned numbers describe the line:

$$\widehat{\mathrm{minutes}} = a + b\,\times\,\mathrm{distance\ (km)}.$$

The small hat on `minutes` means **predicted**, not measured. The model has learned the values of **a** (its starting point) and **b** (how much its prediction changes for each additional kilometer). We will not derive the formula today. Note that the starting point describes the mathematical line at **0 km**, outside much of our observed range; it is not necessarily a meaningful physical estimate for a real zero-distance delivery.

`fit` does not mean the model understood traffic or the delivery business. It adjusted a small number of parameters to match the training examples as well as this method could.

### 4.8 Predict a new delivery

Suppose a new order is **5 km away**. We know that distance before leaving. We do not yet know how many minutes the delivery will actually take.

The library's `.predict(...)` command accepts a table with the *same feature name and structure* used in training. It returns estimated minutes. The `:.1f` part only formats a number to one decimal place.

**Pause and guess:** Will the prediction be exactly the same as our 28-minute handwritten rule?

````python
new_order = pd.DataFrame({"distance_km": [5.0]})
new_prediction = model.predict(new_order)[0]
print(f"Handwritten rule: {fixed_rule(5.0):.1f} minutes")
print(f"Learned model:     {new_prediction:.1f} minutes")
````

The two answers differ because the handwritten rule uses numbers we chose, while the model's numbers came from training examples. The result is a **prediction**, not the actual outcome. We will only know the true trip duration after the delivery is completed.

### 4.9 Does the model make correct predictions on unfamiliar examples?

A single interesting prediction does not establish reliability. The next cell predicts the **seven deliveries that the model did not train on**. We can compare each prediction with the recorded actual outcome.

We call their difference a **prediction error**. For a single delivery, the signed error is `prediction − actual`. A negative value means the model predicted too little time; a positive value means it predicted too much. To measure the size of errors without positive and negative values canceling, we use **absolute error**: `abs(prediction − actual)`.

````python
X_test = testing[["distance_km"]]
testing["predicted_minutes"] = model.predict(X_test)
testing["rule_minutes"] = testing["distance_km"].apply(fixed_rule)
testing["signed_error"] = testing["predicted_minutes"] - testing["actual_minutes"]
testing["absolute_error"] = testing["signed_error"].abs()

testing[["delivery_id", "distance_km", "traffic_score", "actual_minutes",
         "predicted_minutes", "absolute_error"]].round(1)
````

A row with, for example, an **absolute error of 6 minutes** means the estimate missed the real delivery time by six minutes, regardless of whether it was too early or too late. Look for the rows with heavy traffic: some are difficult for a distance-only model because two equally long trips need not take the same time.

We can summarize this first test using **mean absolute error (MAE)**:

$$\mathrm{MAE}=\frac{\text{sum of all absolute errors}}{\text{number of test deliveries}}.$$

MAE is the **average size of the miss** in minutes. Smaller is better *for this particular measure*. We introduce it only as a readable way to summarize errors; model evaluation is studied properly in **L12**.

````python
model_mae = testing["absolute_error"].mean()
rule_mae = (testing["rule_minutes"] - testing["actual_minutes"]).abs().mean()
print(f"Distance-only learned model: {model_mae:.1f} minutes MAE on held-out deliveries")
print(f"Handwritten rule:            {rule_mae:.1f} minutes MAE on held-out deliveries")
````

**Interpret the comparison carefully.** We have compared two methods on the **same seven unseen examples**. A smaller MAE on these records means a smaller average miss *here*; it is not a universal statement that the same method wins in every city or every month. Seven synthetic deliveries are far too few to establish dependable real-world performance.

Also, the rule was not tuned on the training examples, while the model was. That is part of the intended comparison between a hand-written assumption and a learned relationship, not a comprehensive benchmark.

### 4.10 Show what the learned line can and cannot capture

The next plot contains **training observations** (small circles), **held-out test observations** (diamonds), and the model's **predicted line** (black line). The line uses distance only, so it predicts the same time for two trips of equal distance, regardless of traffic.

````python
line_distances = np.linspace(1, 10, 100)  # 100 distances for drawing a smooth line
line_table = pd.DataFrame({"distance_km": line_distances})
line_predictions = model.predict(line_table)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(training["distance_km"], training["actual_minutes"],
           label="Training deliveries", color="#3777a7", s=65, alpha=0.8)
ax.scatter(testing["distance_km"], testing["actual_minutes"],
           label="Held-out deliveries", color="#d18d36", marker="D", s=90)
ax.plot(line_distances, line_predictions, color="#242424", linewidth=2.5,
        label="Learned prediction from distance")
ax.set(title="A learned line: useful trend, imperfect predictions",
       xlabel="Distance at dispatch (km)",
       ylabel="Delivery time (minutes)")
ax.grid(alpha=0.15)
ax.legend()
plt.show()
````

**What to notice:** The black line captures an average trend, not every delivery. Several dots lie above it (the model underestimates their times) and others below it (the model overestimates). The diamonds matter most for our introductory check because those deliveries were **not part of training**.

The line is also an example of a model making an assumption: it treats the relationship between distance and predicted time as straight. Sometimes a straight line is a useful approximation. Other situations need different assumptions or more information. That is why later lectures compare algorithms rather than assuming one is always best.

### 5.3 Data Science, Data Mining, and Machine Learning

**Data Science** is the broader practice of using data to answer questions and create useful understanding. It also includes communicating findings, thinking about data quality, and considering consequences.

**Data Mining** refers to discovering potentially useful patterns or relationships in data. It connects to statistics, databases, and ML. Some data-mining tasks are **predictive** (estimate unknown outcomes); others are **descriptive** (find groups or summarize relationships).

**Machine Learning (ML)** is an approach in which a system uses data or experience to improve how it performs a task rather than relying entirely on individually written rules. A **model** is a learned representation used to map inputs to outputs, identify structure, or guide actions, depending on the learning setting.

### 5.4 Traditional rules versus learning from examples

**Handwritten rules**

- A programmer chooses the relationship.
- The program follows the supplied instructions.
- Strong when reliable domain rules exist and explainability matters.

**Learned models**

- The programmer chooses a learning method and supplies suitable examples.
- Training adjusts the relationship based on patterns in the data.
- Useful when realistic patterns are difficult to describe completely with fixed rules.

**Neither approach is always best:** Both need good assumptions and checking against the real problem.

Both approaches require **human choices**. People still decide which problem to solve, what data to collect, what target means success, and how to evaluate a prediction.

### 5.5 Learning versus memorizing

A model is useful when it can handle **new but relevant examples**. If it merely repeats training examples without discovering a useful relationship, it may fail on the next delivery. Our held-out test is an early way to investigate whether learned behavior transfers to new cases. We will study the balance between fitting examples and handling new ones in **L12**.

A model can also be wrong because it is missing an important feature, the training data are unreliable, its assumptions do not fit the problem, or the conditions have changed. For example, an unexpected road closure may break the usual relationship between distance and time.

### 5.6 Why data quality and evaluation matter

Imagine that ten recorded durations are in **seconds**, but the rest are in **minutes**. A model might learn nonsense unless someone notices the different units. Imagine that a courier's location is updated *after* delivery, but included as if it were known before dispatch. That leaks future information into our prediction. Imagine that all past examples come from a quiet morning while we need to predict rush hour.

These are three different problems: **measurement quality**, **leakage**, and **relevance**. A successful-looking training score cannot rule them out.

A useful evaluation asks more than “Did our code run?” It asks: **How large are the mistakes? On what kind of cases do they occur? What happens on examples the model never saw? What cost would a mistake have?** Our seven-row test is only an introduction; we will learn better evaluation methods at the appropriate points in the course.

## 6. Different questions lead to different kinds of learning

The delivery-time model answers a **numerical prediction question**. But there are many other sensible questions about the very same delivery records. Some need known outcomes for training; others look for structure without an answer column; still others learn by trying actions and observing rewards.

We call these different **learning settings** or **paradigms**. The algorithm should follow the **question and the kind of feedback available**, not the other way around.

### 6.1 Supervised learning: examples with known outcomes

In **supervised learning**, training data contain **features** and a **known target** for each training example. We try to learn a relationship that helps estimate a target for a new observation.

There are two important tasks:

- **Regression:** predict a **numerical quantity** such as 34 minutes, electricity consumption, or tomorrow's temperature. Our first delivery example is regression.
- **Classification:** predict one of a set of **categories** such as “late/on time,” “spam/not spam,” or a product type. A classifier may also produce scores or probabilities, but its central question concerns a category.

A numerical code used to stand for a category is not automatically a regression target. What matters is **what the output means**, not merely whether it is stored as a number.

We will learn regression in **L11–L12**, classification in **L13–L18**, and classification evaluation immediately in **L14**.

**Check:** Is predicting the number of passengers on a bus regression or classification? It is a numerical/count prediction, so the natural first answer is regression, even though the count is a whole number. A separate question like “crowded/not crowded” would be classification.

### 6.2 Unsupervised learning: structure without a known target

Sometimes we have many observations but no column telling us the answer. **Unsupervised learning** looks for structure in input data without training on target labels for the task.

Two examples are:

- **Clustering:** find groups of observations that are similar according to chosen features. A delivery service might find groups of short, medium, and long trips with different traffic conditions **without being told those groups in advance**.
- **Anomaly detection:** find observations that behave unusually compared with expected patterns. One trip could take much longer than is typical given its conditions. Unusual does **not** automatically mean fraudulent or incorrect; it is a case worth investigating.

These are different objectives: clustering emphasizes **groups**; anomaly detection emphasizes **departures from expected behavior**. We will study clustering in **L19–L20** and Isolation Forest for anomalies in **L21**.

### 6.3 Semi-supervised learning: many examples, few known labels

A team may have 10,000 photos but only 200 have been reviewed and labeled with what they show. **Semi-supervised learning** uses both the small labeled set and a larger unlabeled set to improve learning, under assumptions about how the examples relate.

Expected outcome: a model that can benefit from unlabeled examples, rather than ignoring them. But extra unlabeled data are not always helpful if they are very different or misleading. We will explore this in **L22**.

### 6.4 Active learning: decide what should be labeled next

Suppose a human expert can review only 20 delivery complaints. It may be wasteful to label 20 nearly identical cases. **Active learning** asks the system to select **which unlabeled examples would be most useful to ask a human or another labeling process about**.

Expected outcome: better learning with fewer expensive labels, by choosing informative labeling requests. This is distinct from semi-supervised learning: **semi-supervised learning uses available unlabeled data in training; active learning chooses which examples should receive labels**. They can also be combined. We revisit both in **L22**.

### 6.5 Reinforcement learning: an agent learns from rewards

Consider a robot learning a route through a small maze. At each step, it observes its **state** (such as its current position), chooses an **action** (move up, down, left, or right), and receives a **reward** or penalty from the **environment**. It wants to learn choices that produce good results over time.

This is **reinforcement learning (RL)**. It is different from ordinary supervised learning: the learner is not given a table of correct actions for every situation. Instead, it receives feedback connected to the consequences of its own actions, sometimes after several steps.

Expected outcome: an **action strategy**, often called a **policy**, telling the agent how to act in different states. We will implement an approachable form using **Q-tables** in **L23–L24**.

### 6.6 Neural networks: a model family, not another kind of label

An **artificial neural network** is a learnable model composed of connected units that transform inputs. It can learn a relationship between features and targets, or support other learning settings. **Deep learning** usually uses networks with multiple learned layers.

A neural network predicting delivery time is still doing **supervised regression**. A neural network choosing robot actions from rewards can take part in **reinforcement learning**. This is why “neural networks” should not be listed as an independent paradigm equal to “supervised learning.” We focus on the neural-network model family in **L25–L30**.

### 6.7 Compare the learning settings

**With known answers → Supervised learning**

- **Regression (L11–L12):** numerical training target → estimate delivery minutes.
- **Classification (L13–L18):** category training target → predict late / on time.

**Without provided target labels → Unsupervised learning**

- **Clustering (L19–L20):** discover groups of similar deliveries.
- **Anomaly detection (L21):** identify unusual cases for investigation.

**When labeling examples is difficult**

- **Semi-supervised learning (L22):** use a few labeled photos together with many unlabeled ones.
- **Active learning (L22):** choose which observations a person should label next.

**When an agent learns by acting**

- **Reinforcement learning (L23–L24):** learn useful actions by receiving rewards, such as when navigating a maze.

**A model family, not an alternative setting**

- **Neural networks / deep learning (L25–L30):** learn flexible representations; can be used for several different learning tasks.

> **Choose the question first.** The inputs, targets, feedback, algorithms, and evaluation methods come afterward.

## 7. One dataset, four different questions

Let's return to the deliveries. Changing the learning **objective** changes what kind of problem we are solving, even if some columns stay the same.

1. **Regression:** “How many minutes will this delivery take?” Training target: actual minutes, a number.
2. **Classification:** “Will this delivery arrive after its promised time?” Training target: late or on time, a category known after delivery.
3. **Clustering:** “Are there groups of deliveries with similar distances and traffic conditions?” There is **no provided group label**; we seek structure in the features.
4. **Anomaly detection:** “Which completed trips took unusually long compared with typical trips under similar conditions?” We flag unusual observations, **not** automatically confirmed mistakes or fraud.

These are not four styles of the same output. They are **four different questions**, possibly requiring different data and different kinds of feedback.

### 7.1 Make a late/on-time target for teaching

Imagine the service sets a **promised travel-time allowance** at dispatch based on known conditions. Our toy allowance will be `9 + 3.5 × distance + 6 × traffic_score` minutes. This is a made-up operational rule and **not a learned promise**. We determine whether the delivery was late only after we see the **actual time**.

The promise is available at dispatch. The **late label** is not. It is safe to train a classifier on historical late labels, then use only dispatch-time features for new deliveries. It would be leakage to use the completed delivery time to classify that *same* delivery before it finishes.

````python
deliveries["promised_minutes"] = (
    9 + 3.5 * deliveries["distance_km"] + 6 * deliveries["traffic_score"]
)
deliveries["late"] = deliveries["actual_minutes"] > deliveries["promised_minutes"]

print("Counts of completed deliveries by historical late/on-time status:")
print(deliveries["late"].map({False: "On time", True: "Late"}).value_counts())
deliveries[["delivery_id", "distance_km", "traffic_score",
            "promised_minutes", "actual_minutes", "late"]].head(10).round(1)
````

**Read the result:** `late` is either `True` or `False`, a category meaning the recorded duration exceeded the promise. Although the label comes from comparing two numbers, predicting `late` for a **future** delivery would be a **classification** task, not ordinary regression. We have created historical labels; **we have not trained a classifier yet**. That happens beginning in **L13**.

The code uses `>` to mean “greater than,” and `.value_counts()` counts how many rows belong to each category.

### 7.2 A visual comparison of four objectives

The next figure uses the same observations in two deliberately different ways:

- **Left:** observations colored by their **known historical late label** (after the trip). A classifier would use such labels for training and try to predict them from information available beforehand.
- **Right:** observations colored only by **traffic at dispatch**, without any target label. The patterns might motivate a clustering investigation, but assigning colors from an existing traffic column **is not itself a clustering algorithm**.

Here the axes are distance and actual time so we can inspect historical trips. **For a real clustering system operating before dispatch, we would not use the completed time as an input**; we would use dispatch-time information such as distance and traffic. The chart is exploratory, not a deployed model.

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.7), sharex=True, sharey=True)
for state, name, color in [(False, "On time", "#2b7a78"), (True, "Late", "#b44a52")]:
    subset = deliveries[deliveries["late"] == state]
    axes[0].scatter(subset["distance_km"], subset["actual_minutes"],
                    color=color, s=70, alpha=.86, label=name)
for traffic in [0,1,2]:
    subset = deliveries[deliveries["traffic_score"] == traffic]
    axes[1].scatter(subset["distance_km"], subset["actual_minutes"],
                    color=traffic_colors[traffic], s=70, alpha=.86,
                    label=traffic_names[traffic])
for ax, title in zip(axes, ["Known outcome: was it late?", "Describing trips by known traffic"]):
    ax.set(title=title, xlabel="Distance (km)", ylabel="Actual minutes")
    ax.grid(alpha=.12)
    ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
````

**Interpretation:** In the left panel, the colors describe a **completed outcome**. In the right panel, they describe a **feature that was already known**. Different colors do not automatically mean the points form natural clusters. Also, a “late” label depends on the promised time—not just the absolute length of the delivery. A 45-minute trip could be on time for one promise and late for another.

### 7.3 What would make a trip an anomaly?

An unusually long completed trip could deserve investigation even if it was not officially late. Suppose there was a road closure but the customer received a generous promised window. Anomaly detection might notice a strange travel duration; classification might still say **on time**. Conversely, a delivery could be only slightly unusual but **late** because its promise was very tight.

This is why the **definition of the question** comes before the model. Anomalies depend on context and the chosen features. A high duration alone does not prove that a delivery is an anomaly, and an anomaly flag is not an explanation of its cause.
